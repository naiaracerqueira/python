## Python Decorators

Decoradores são funções que adicionam comportamentos a outra função, sem modificá-la.

### Caso 1: Wrapper vazio

A função _decorator_ recebe a função como argumento e retorna essa mesma função com um comportamento adicionado. Colocamos a função wrapper dentro dela para que ele só seja chamado quando a função decorada é chamada.

Nesse caso, a função _add_sprinkles_ é a função _decorator_, que recebe a função get_ice_cream() como func. A função _wrapper_ é a função interna do _decorator_ que recebe os mesmos argumentos que a função inicial.

Nesse caso a função _decorator_ não faz nada. Como o decorator não faz nada, ele só printa a mensagem original:

```python
def add_sprinkles(func):
    def wrapper(): # só é chamado quando a função decorada é chamada (get_ice_cream())
        func()
    return wrapper

@add_sprinkles
def get_ice_cream():
    print(f"Here is your ice cream 🍨")

get_ice_cream()
```

### Caso 2: Wrapper com print

Nesse caso, começamos a criar as funções _decorator_: add_sprinkles e add_fudge. Ambas recebem a função original como argumento (func), fazem um print e chamam a função original de volta.

```python
def add_sprinkles(func): # Recebe a função original como argumento
    def wrapper():
        print("*You add sprinkles 🎊*")
        func() #Chama a função original (get_ice_cream())
    return wrapper

def add_fudge(func):
    def wrapper():
        print("*You add fudge 🍫*")
        func()
    return wrapper

@add_sprinkles
@add_fudge
def get_ice_cream():
    print(f"Here is your ice cream 🍨")

get_ice_cream()
```

### Caso 3: função original com argumentos

Se eu preciso passar argumentos para a função original, eu preciso passar os mesmos argumentos para a função wrapper. Ou seja, wrapper tem que receber os argumentos da função original e isso pode variar de função decorada para função decorada, certo?

A forma mais fácil e simples de fazer isso, é com *args, **kwargs, que permitem que ele receba qualquer função com qualquer número de argumentos.

Nesse caso, temos apenas um argumento: posso passar como "vanilla" (args recebe) ou number=2, flavor="vanilla" (kwargs recebe)

```python
def add_sprinkles(func):
    def wrapper():
        print("*You add sprinkles 🎊*")
        func()
    return wrapper

def add_fudge(func):
    def wrapper():
        print("*You add fudge 🍫*")
        func()
    return wrapper

@add_sprinkles
@add_fudge
def get_ice_cream(flavor):
    print(f"Here is your {flavor} ice cream 🍨")

get_ice_cream("vanilla")
```

```python
def add_sprinkles(func):
    def wrapper(*args, **kwargs): # a wrapper tem que receber os argumentos da função original
        print("*You add sprinkles 🎊*")
        func(*args, **kwargs)
    return wrapper

def add_fudge(func):
    def wrapper(*args, **kwargs):
        print("*You add fudge 🍫*")
        func(*args, **kwargs)
    return wrapper

@add_sprinkles
@add_fudge
def get_ice_cream(number, flavor):
    print(f"Here is your {number} {flavor} ice cream 🍨")

get_ice_cream(number=2, flavor="vanilla")
```

### Caso 4: Função com retorno

Quando a função original retorna algo, a wrapper tem que retornar o mesmo valor para que o resultado seja o mesmo

```python
def add_sprinkles(func):
    def wrapper(*args, **kwargs):
        print("*You add sprinkles 🎊*")
        return func(*args, **kwargs) # tem que ter o retorno
    return wrapper

def add_fudge(func):
    def wrapper(*args, **kwargs):
        print("*You add fudge 🍫*")
        return func(*args, **kwargs) # tem que ter o retorno
    return wrapper

@add_sprinkles
@add_fudge
def get_ice_cream(flavor):
    print(f"Here is your {flavor} ice cream 🍨")
    retorno = "I love ice cream!"
    return retorno

retorno = get_ice_cream("vanilla")
print(retorno)
```
