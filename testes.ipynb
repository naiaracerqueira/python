```python
lista = [("lng=-47.802664728268745, lat=-15.657013854445248"), 
         ("lng=-47.75042666463918, lat=-15.733476257312468"), 
         ("lng=-47.81892460800397, lat=-15.69134186360214"), 
         ("lng=-47.86033150700153, lat=-15.656099266599636"),
         ("lng=-47.807670409758806, lat=-15.653245748680474")]
```

```python
lista_nova = []

for tupla in lista:
    longitude = re.findall("lng=(.*),", tupla)[0]
    latitude = re.findall("lat=(.*)", tupla)[0]
    lista_nova.append(longitude)
    lista_nova.append(latitude)

lista_nova
```
